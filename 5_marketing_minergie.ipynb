# Prospection dans les communes *Eco Friendly*

Minergie, "le label suisse pour le confort, l'efficacité et la protection du climat" met en open data les bâtiments qui ont été construits ou rénovés, puis certifiés selon ses standards.

Vous faites du marketing pour un énergéticien qui souhaite proposer des forfaits d'électricité "verte".
Vous souhaitez cibler un persona *Eco Friendly* avec des publicités dans les communes (bord de route, arrêt de bus, *etc*) qui ont le plus de bâtiments Minergie.

Vous avez pour cela 2 fichiers :
- `ogd91_minergiegebaeude_pro_gemeinde.csv` : les communes avec le nombre de bâtiments [Minergie](https://opendata.swiss/fr/dataset/anzahl-minergie-gebaude-in-gemeinden/)
- `communes-compressé.geojson` : les limites topologiques des communes compressées avec une moindre résolution
