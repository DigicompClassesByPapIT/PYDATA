# Prédiction du prix du BTC à l'aide d'un modèle LSTM avec TensorFlow et visualisation avec Plotly

[`yfinance`](https://github.com/ranaroussi/yfinance) est un module Python qui permet de télécharger des données financières à partir de Yahoo Finance.
Il permet d'accéder à l'historique de nombreuses valeurs sans payer ou s'inscrire.

Tensorflow est une bibliothèque de machine learning/ Deep Learning open source développée par Google, voir https://www.tensorflow.org/guide/keras?hl=fr.
Keras est une API haut niveau de Tensorflow pour construire et entraîner des modèles de Deep Learning. Ce module met à disposition des réseaux de neurones récurents (RNN) comme le modèle LSTM (Long Short-Term Memory) qui sont adaptés pour les séries temporelles.

À l'aide de ces 2 librairies, nous allons prédire le prix du Bitcoin (BTC) de demain à partir de ses données historiques. Le résultat sera visualisé avec Plotly.

Création du dataset du BTC moyen journalier du BTC (la moyenne entre le prix d'ouverture et de fermeture) sur les 90 derniers jours.

Normalisation et création du modèle LSTM pour prédire le prix du BTC de demain à partir des données historiques.

Visualisation du prix moyen journalier du BTC et de la prédiction pour demain avec Plotly.
