# Plotly

Visualiser des données stockée dans le fichier `df_orders.parquet` avec la librairie Plotly.

1. Exporter les données du fichier `df_orders.parquet` dans le DataFrame `df_orders`.

2. Créer avec Plotly la courbe du nombre de commandes en fonction du prix, la courbe en turquoise #1abc9c et les points en orange. Ajouter une infobulle affichant le prix et le nombre de commandes pour chaque point.

3. Créer un graphique, de type camembert, avec Plotly, représentant la répartition des commandes des 10 premiers restaurants.

4. À l'aide de Plotly Express, créer un graphique en barres représentant le nombre de commandes par trimestre sur l'année 2019. Ajouter un titre et des étiquettes aux axes.
