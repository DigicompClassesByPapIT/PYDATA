## Import des packages

```python
import numpy as np
import pandas as pd
import plotly.express as px
```

## Import et vérification du jeu de données

```python
df = pd.read_csv('./salary-data.csv')
```

```python
df.head()
```

## removing unnecessary columns

```python
df.drop(columns=["Unnamed: 0"], axis=1, inplace=True)
```

## Checking for Null Values

```python
df.isnull().sum()
```

## Check for correlation

```python
df.corr()
```

## visualizing data points using plotly express

```python
plot = px.scatter(
    df,
    x='Experience Years',
    y='Salary',
    title='Salary vs YearsExperience')
plot.show()
```

## independent and dependent features

```python
X = df[['Experience Years']]  ## avoiding series datatype for the input features
y = df['Salary']
```

## Performing train test split

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

## Standardization

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)  ## avoiding data leakage
```

```python
## Applying Linear Regression
```

```python
from sklearn.linear_model import LinearRegression

regression = LinearRegression()
```

```python
regression.fit(X_train, y_train)
```

```python
print("Coefficient or slope: ", regression.coef_)
print("Intercept: ", regression.intercept_)
```

## Plotting Training data for best fit line using plotly

```python

plot = px.scatter(
    x=X_train.flatten(),
    y=y_train,
    title='Salary vs YearsExperience',
    labels={'x': 'Experience Years', 'y': 'Salary'}
)
plot.add_scatter(x=X_train.flatten(), y=regression.predict(X_train), mode='lines', name='Best Fit Line')
plot.show()
```

## Prediction

```python
y_pred = regression.predict(X_test)
```

## Performance Metrics

### MAE, RSE, RMSE

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mse)
print("MSE: ", mse)
print("MAE: ", mae)
print("RMSE ou écart quadratique moyen: ", rmse)
```

### R-square

```python
from sklearn.metrics import r2_score

score = r2_score(y_test, y_pred)
print("R2: ", score)

print("R2 ajusté: ", (1 - (1 - score) * (len(y_test) - 1) / (len(y_test) - X_test.shape[1] - 1)))
```
